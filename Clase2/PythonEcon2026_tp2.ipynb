## Contexto: el modelo SIR

Durante la pandemia de COVID-19, una pregunta central para gobiernos y hospitales era cuánto se podía *aplanar la curva* de contagios. En este TP usamos [el modelo SIR](https://es.wikipedia.org/wiki/Modelo_SIR) para simular los contagios en cada comuna de la Ciudad de Buenos Aires. El modelo clasifica a las personas en:

- **S — Susceptibles:** pueden contagiarse.
- **I — Infectadas:** pueden contagiar. Las dividimos en **I0** (primera semana de la enfermedad) e **I1** (segunda semana, próximas al alta).
- **R — Recuperadas:** ya no contagian ni se contagian.

Las personas pasan de S a I según una **tasa de contagio** y de I a R al cabo de dos semanas. El modelo está simplificado a propósito: el objetivo es practicar funciones, loops, listas y diccionarios.

```python
import csv
import random
```

```python
# Este bloque de código importa los datos. No tienen que modificar nada.
# Datos reales de la Ciudad de Buenos Aires, descargados de:
# https://cdn.buenosaires.gob.ar/datosabiertos/datasets/salud/casos-covid-19/casos_covid19.csv
# Corran este notebook en la misma carpeta donde guardaron los archivos CSV que 
# les entregamos junto con este TP.

ARCHIVOS = ['infectados0.csv', 'infectados1.csv', 'recuperados.csv']

def obtengo_infectados_por_comuna(archivos=ARCHIVOS):
    '''
    Lee los tres archivos de casos y devuelve una lista de personas por archivo.

    Input:
        archivos (list): nombres de los tres CSV, en este orden: infectados en
            la última semana, infectados hace una semana, recuperados.

    Output:
        tupla con tres listas (infectados0, infectados1, recuperados).
        Cada lista contiene una tupla por persona: (comuna, genero, edad),
        donde los tres valores son strings, tal como vienen en el archivo.
    '''
    listas_personas = []

    for archivo in archivos:
        with open(archivo, 'r', encoding='utf-8') as f:
            filas = csv.reader(f)

            personas = []
            for fila in filas:
                personas.append((fila[0], fila[1], fila[2]))

        listas_personas.append(personas)

    infectados0, infectados1, recuperados = listas_personas

    return (infectados0, infectados1, recuperados)
```

#### Ejercicio 6
La función `es_infectado()` determina si una persona susceptible se contagia. Recibe la comuna, la posición de la persona y la tasa de contagio.

Si al menos uno de sus dos vecinos está infectado (`'I0'` o `'I1'`), le asignamos a la persona un valor aleatorio de "inmunidad" con `random.random()`. Si ese valor es **menor** que la tasa de contagio, se contagia; si no, no.

Notas:
- La primera posición de una lista es la 0.
- La comuna es un **anillo**: la primera y la última persona de la lista son vecinas.

**Antes de completar:** expliquen en una oración por qué el código no necesita un `if` especial para la persona en la posición 0. (Pista: ¿qué es `comuna[-1]`?)

*Respuesta:*

```python
def es_infectado(comuna, posicion, tasa_de_contagio):
    '''
    Determina si una persona susceptible se contagia esta semana.

    Inputs:
        comuna (list): el estado de cada persona al comienzo de la semana
        posicion (int): la posición de la persona analizada
        tasa_de_contagio (float): probabilidad de contagiarse si al menos
            un vecino está infectado

    Output:
        True si la persona se contagia, False si no.
    '''
    pos_izq = posicion - 1
    pos_der = posicion + 1

    # Si es la última persona, su vecino de la derecha es la persona 0
    if posicion == len(comuna) - 1:
        pos_der = 0

    if comuna[pos_izq] in ('I0', 'I1') or comuna[pos_der] in ('I0', 'I1'):
        if random.random() < tasa_de_contagio:
            return None   # COMPLETEN: ¿qué debería devolver acá?

    return None           # COMPLETEN: ¿y acá?
```

```python
# Verificación (con tasa 1 o 0 el resultado no depende del azar)
assert es_infectado(['I0', 'S', 'S'], 1, 1) == True,  "Vecino izquierdo infectado: debería contagiarse"
assert es_infectado(['S', 'S', 'S', 'S'], 1, 1) == False, "Sin vecinos infectados: no se contagia"
assert es_infectado(['S', 'S', 'I1'], 0, 1) == True,  "La posición 0 es vecina de la última"
assert es_infectado(['I0', 'S', 'S', 'S'], 3, 1) == True, "La última posición es vecina de la 0"
assert es_infectado(['I0', 'S', 'I1'], 1, 0) == False, "Con tasa 0 nadie se contagia"
print("✔ Ejercicio 6 OK")
```

Simulen el paso de una semana con la función `simular_una_semana()`. Al cabo de una semana:

- a quienes estaban en `'S'` les aplicamos `es_infectado()` para saber si pasan a `'I0'` o siguen en `'S'`.

La función recibe la comuna y la tasa de contagio, y **devuelve una lista nueva** con el estado de cada persona al comienzo de la semana siguiente.

**Importante:** el contagio se decide mirando la comuna **al comienzo de la semana** (`comuna_s1`), no la lista que están construyendo. ¿Por qué importa esto? (Lo van a necesitar en el Ejercicio 10.)

```python
def simular_una_semana(comuna_s1, tasa_de_contagio):
    '''
    Avanza la simulación una semana.

    Inputs:
        comuna_s1 (list): el estado de cada persona al comienzo de la semana t
        tasa_de_contagio (float): probabilidad de contagiarse si al menos
            un vecino está infectado

    Output:
        comuna_s2 (list): una lista NUEVA con el estado de cada persona al
            comienzo de la semana t+1
    '''
    comuna_s2 = []

    # i es la posición y val el estado de cada persona
    for i, val in enumerate(comuna_s1):
        if val == 'R':
            comuna_s2.append('R')

        elif val == 'I0':
            pass   # COMPLETEN

        # COMPLETEN: el caso 'I1'

        elif val == 'S':
            # Usamos es_infectado sobre la comuna al COMIENZO de la semana
            if es_infectado(comuna_s1, i, tasa_de_contagio):
                comuna_s2.append('I0')
            else:
                comuna_s2.append('S')

    return comuna_s2
```
