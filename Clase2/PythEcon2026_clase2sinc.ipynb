## 3. Funciones: scope y mutabilidad

Hasta ahora dijimos que las variables creadas dentro de una función son **locales**.

Pero hay otra pregunta distinta:

> ¿Qué pasa cuando una función recibe un objeto mutable, como una lista o un diccionario, y lo modifica?

Vamos a separar ambas ideas.


### 3.1 Predicción 1: una variable local

**No ejecuten todavía.** ¿Qué creen que imprimirá?


````python
numero = 10

def sumar_diez(x):
    x += 10

sumar_diez(numero)

print(numero)
````

**Pregunta:** ¿por qué `numero` sigue valiendo lo mismo aunque dentro de la función aparezca `x = 20`?


### 3.2 Predicción 2: una lista

**Otra vez: predigan antes de ejecutar.** ¿Qué creen que imprimirá ahora?


````python
numeros = [10, 20]

def alargar_lista(x):
    x.append(30)

alargar_lista(numeros)

print(numeros)
````

### Idea clave

Estos dos ejemplos NO muestran exactamente el mismo fenómeno.

- `x = 20` hace que el nombre local `x` pase a referirse a otro objeto.
- `x.append(20)` **modifica el objeto (la lista) que ya existía**.

Por eso una función puede cambiar una lista o un diccionario que recibió incluso aunque no tenga `return`.

### 3.3 Volvamos al ejemplo económico

¿Qué esperan que ocurra con `paises_info` después de llamar a esta función?


````python
# Reiniciamos el diccionario para ver el efecto con claridad
paises_info = {
    'argentina': {'cod': 'ARG', 'pob': 44939, 'gdp': 1031215},
    'bolivia':   {'cod': 'BOL', 'pob': 11513, 'gdp': 104609},
    'brasil':    {'cod': 'BRA', 'pob': 211050, 'gdp': 3220373},
    'chile':     {'cod': 'CHL', 'pob': 18952, 'gdp': 476738},
    'paraguay':  {'cod': 'PRY', 'pob': 7045, 'gdp': 93062},
    'uruguay':   {'cod': 'URY', 'pob': 3462, 'gdp': 77732}
}

def agrega_gdp_pc(dic_compuesto):
    """Agrega la llave 'gdp_pc' a cada diccionario interno.
    Input:
    dic_compuesto (dic): recibe un diccionario que tiene como llave el nombre
                         del pais y como valor un diccionario con los valores
                         de su poblacion, gdp y código
    """
    for pais, datos in dic_compuesto.items():
        datos["gdp_pc"] = gdp_per_capita(datos["gdp"], datos["pob"])
````

# 7. Kernel, memoria y reproducibilidad

Durante un notebook podemos:

- crear variables;
- redefinir funciones;
- ejecutar celdas fuera de orden;
- borrar código que ya dejó objetos guardados en memoria.

Por eso, que una celda funcione **ahora** no garantiza que el notebook funcione para alguien que los abre de cero.

El kernel conserva información en memoria, por lo que una buena práctica para verificar reproducibilidad es usar ***Restart & Run All*** para comprobar que un notebook es reproducible:

## Antes de entregar

Usen:

El notebook debería correr de arriba hacia abajo sin depender de “cosas que quedaron en memoria”.
