### Cómo armar el `requirements.txt`

Es la lista de librerías que hay que instalar para que su notebook corra en otra computadora. Se instala todo junto con `pip install -r requirements.txt`.

**Reglas**

- Una librería por línea, con la versión exacta: `pandas==3.0.5`. Las líneas que empiezan con `#` son comentarios.
- Incluyan **todas las librerías externas que importa el notebook**, más `ipykernel` (para poder usar el entorno desde VS Code o Jupyter).
- **No incluyan** las librerías que vienen con Python (`re`, `os`, `json`, `math`, `datetime`, `random`): no se instalan con `pip`.
- **No incluyan** Python: la versión de Python se anota en el `README.md`.
- **No peguen la salida completa de `pip freeze`**: trae cientos de paquetes y, a veces, rutas de su computadora que no existen en otra.
- Usen el **nombre con el que se instala**, que no siempre es el del `import`: `import sklearn` se instala como `scikit-learn`, y `import bs4` como `beautifulsoup4`.

**Ejemplo** (las versiones son ilustrativas: copien las que les imprime la celda de verificación del ejercicio 0):

    # TP4 · Grupo X
    numpy==2.5.3
    pandas==3.0.5
    matplotlib==3.11.2
    seaborn==0.13.2
    rapidfuzz==3.14.6
    ipywidgets==...
    ipykernel==...

    Probado con Python 3.12.

**(4)** Escriban una función `parsear_domicilio(texto)` que reciba un texto **ya normalizado** y devuelva un diccionario con tres claves, `"barrio"`, `"manzana"` y `"casa"`. Reglas:

- **Barrio:** solo si el texto empieza con `B°`. Es lo que sigue a `B°`, hasta el primer marcador de manzana o de casa, o hasta el primer `;`.
- **Manzana:** solo si está **marcada** con `M`, `MZ`, `MZA`, `MZNA` o `MN` (seguido o no de `.`, `:` o espacios). Es el código que sigue al marcador: letras, números o una combinación (`G`, `LL`, `40`, `J10`).
- **Casa:** solo si está **marcada** con `C`, `CS`, `CSA` o `CASA`. Es el número que sigue al marcador, **sin ceros a la izquierda** (`"03"` pasa a `"3"`). Una casa `0` no es un dato válido: va como `None`.
- Lo que no se puede determinar va como `None`. **Un número suelto, sin marcador, no es ni manzana ni casa.**

Usen grupos con nombre, `(?P<nombre>...)`, y `\b` donde haga falta. Empiecen por el primer domicilio, `"B° SAN LUIS XV(15);C: 03;;M: G;"`, que debe devolver `{"barrio": "SAN LUIS XV(15)", "manzana": "G", "casa": "3"}`.

La verificación prueba la función con 14 casos. Los profesores tenemos **otros casos de prueba** que no están acá: una función que solo pasa estos 14 porque está hecha a medida no va a pasar los otros.

Nos devolvió el código de la celda de abajo. **Corre sin errores y produce una tabla razonable**, pero tiene **tres errores** que hacen que el resultado esté mal.

```python
import re
import pandas as pd
from rapidfuzz import process, fuzz

ia = pd.read_csv("domicilios.csv", sep=";", encoding="latin1", dtype=str, keep_default_na=False)
ia["texto"] = (ia["calle3"] + ";" + ia["numero"] + ";" + ia["depto_normal"] + ";"
               + ia["manzana_normal"] + ";" + ia["lote_normal"])
ia["texto"] = ia["texto"].str.upper().str.strip()

# extraemos barrio, manzana y casa con grupos con nombre
patron = r"B°\s*(?P<barrio>.*)\s*M\w*[.: ]*(?P<manzana>\w+)\s*C\w*[.: ]*(?P<casa>\d+)"
ia = ia.join(ia["texto"].str.extract(patron))

# unificamos los nombres con fuzzy matching
referencia_ia = pd.read_csv("barrios_referencia.csv")["barrio"].tolist()
def unificar(nombre):
    if pd.isna(nombre):
        return None
    return process.extractOne(nombre.strip(), referencia_ia, scorer=fuzz.WRatio)[0]

ia["barrio_unificado"] = ia["barrio"].apply(unificar)
conteo_ia = ia.dropna(subset=["manzana", "casa"]).groupby("barrio_unificado").size().sort_values(ascending=False)
conteo_ia.head(10)
```
