## 2. NumPy: mínimos cuadrados a mano

En el video quedaron pendientes tres cosas: fijar una **semilla**, generar números de una distribución **normal** y el **producto matricial**. Las juntamos en un ejemplo que conocen bien: estimar por MCO

$$\hat\beta = (X'X)^{-1}X'y$$

Simulamos datos donde sabemos la verdad (constante 5, pendiente 0,8) y vemos si la recuperamos.

````python
np.random.seed(50)   # semilla: los números "aleatorios" serán siempre los mismos

n = 500
educ = np.random.randn(n) * 3 + 12   # normal con media 12 y desvío 3
u = np.random.randn(n) * 2           # error: normal con media 0 y desvío 2
salario = 5 + 0.8 * educ + u

# X: una columna de unos (la constante) y la columna educ
X = np.column_stack([np.ones(n), educ])
print(X.shape)

# beta = (X'X)^(-1) X'y
beta = np.linalg.inv(X.T @ X) @ X.T @ salario
print(beta)
````

- `.T` es la traspuesta, `@` el producto matricial y `np.linalg.inv` la inversa.
- `np.random.randn(n)` devuelve normales estándar; multiplicar por el desvío y sumar la media las reescala. Es broadcasting.
- Con la misma semilla, todos obtenemos exactamente los mismos números.

## 5. Matplotlib: fechas y primer gráfico

Matplotlib es la librería base para graficar. Vamos a graficar la potencia total del país a lo largo del tiempo. Para eso necesitamos dos cosas: que pandas entienda `periodo` como una **fecha** y una tabla con **un valor por mes**.

**Fechas.** `periodo` es texto con formato día/mes/año. Miren qué pasa si no le decimos el formato a pandas:

````python
print(df["periodo"].iloc[0])
print(pd.to_datetime(df["periodo"].iloc[0]))                             # ¿10 de enero?
print(pd.to_datetime(df["periodo"].iloc[0], format="%d/%m/%Y %H:%M"))    # 1 de octubre
````

En este archivo las fechas son ambiguas: `01/10/2015` puede leerse como 1 de octubre o como 10 de enero. Si no indicamos el formato, pandas las interpreta mal y **no da ningún error**. Siempre indiquen el formato.

````python
df["fecha"] = pd.to_datetime(df["periodo"], format="%d/%m/%Y %H:%M")
print(df["fecha"].dtype)
````

````python
total = df.groupby(by="fecha").agg({"potencia_instalada_mw": "sum"}).reset_index()
total.head()
````

**El gráfico.** `fig` es la hoja y `ax` es el gráfico. Todo lo que queremos cambiar del gráfico se lo pedimos a `ax`.

````python
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(total["fecha"], total["potencia_instalada_mw"], marker="o", markersize=3)
ax.set_title("Potencia instalada total")
ax.set_xlabel("Período")
ax.set_ylabel("Potencia instalada (MW)")
ax.tick_params(axis="x", rotation=45)
plt.show()
````

**6.3** En ese gráfico las renovables casi no se ven, porque parten de un nivel muy bajo. Para comparar crecimientos sirve un **índice base 100**. Primer paso: armen `inicial`, una tabla con la potencia de cada fuente en el primer mes, con las columnas `fuente_generacion` y `pot_inicial`.

<details><summary>Pista</summary>

Filtren `por_fuente` por la fecha mínima (`por_fuente["fecha"].min()`), quédense con dos columnas y renombren una. Para renombrar: `rename(columns={"viejo": "nuevo"})`.

````python
# Tu código acá
primera_fecha = por_fuente["fecha"].min()
inicial = por_fuente[por_fuente["fecha"] == primera_fecha][["fuente_generacion", "potencia_instalada_mw"]]
inicial = inicial.rename(columns={"potencia_instalada_mw": "pot_inicial"})
inicial
````

**6.4** Unan `por_fuente` con `inicial` y creen la columna `indice`: la potencia de cada mes dividida por la inicial de esa fuente, por 100. Verifiquen que el resultado tenga la misma cantidad de filas que `por_fuente`. Después grafiquen el índice de las cuatro fuentes.

Para cerrar, respondan en una celda de texto, en dos oraciones: ¿qué fuente creció más en términos relativos? ¿Y cuál sumó más MW? ¿Por qué las dos respuestas son distintas?

<details><summary>Pista</summary>

La clave de la unión es `fuente_generacion`. Para graficar las cuatro fuentes sin copiar y pegar cuatro veces, recorran `por_fuente["fuente_generacion"].unique()` con un `for`.
</details>

````python
# Tu código acá
por_fuente = por_fuente.merge(inicial, on="fuente_generacion", how="left")
por_fuente["indice"] = por_fuente["potencia_instalada_mw"] / por_fuente["pot_inicial"] * 100
print(len(por_fuente))   # sigue teniendo 152 filas

fig, ax = plt.subplots(figsize=(9, 4))
for fuente in por_fuente["fuente_generacion"].unique():
    datos = por_fuente[por_fuente["fuente_generacion"] == fuente]
    ax.plot(datos["fecha"], datos["indice"], label=fuente)
ax.set_title("Potencia instalada por fuente (octubre de 2015 = 100)")
ax.set_xlabel("Período")
ax.set_ylabel("Índice")
ax.legend()
plt.show()
````

**6.5 (desafío)** Armen una figura con dos gráficos, uno al lado del otro: a la izquierda los niveles en MW de las cuatro fuentes y a la derecha el índice. Busquen cómo usar `plt.subplots(1, 2)`.
