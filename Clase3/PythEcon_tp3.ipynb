### Parte C: auditar código generado por IA

Le pedimos a una IA lo siguiente: *«Con el archivo potencia_instalada_mod.csv, calculá qué porcentaje de la potencia instalada total corresponde a fuentes renovables en cada mes, y graficalo»*. Nos devolvió el código de la celda de abajo. **Corre sin errores y produce un gráfico**, pero tiene **tres errores** que hacen que el resultado esté mal.

````python
import pandas as pd
import matplotlib.pyplot as plt

ia = pd.read_csv("potencia_instalada_mod.csv", encoding="latin1", sep="|")
ia = ia[["periodo", "agente", "region", "fuente_generacion", "tecnologia", "potencia_instalada_mw"]]

# eliminamos registros duplicados
ia = ia.drop_duplicates()

# convertimos el período a fecha
ia["fecha"] = pd.to_datetime(ia["periodo"])

# potencia por fecha y fuente
por_fuente_ia = ia.groupby(["fecha", "fuente_generacion"])["potencia_instalada_mw"].mean().reset_index()

# total por fecha y participación de las renovables
total_ia = por_fuente_ia.groupby("fecha")["potencia_instalada_mw"].sum().reset_index()
total_ia = total_ia.rename(columns={"potencia_instalada_mw": "total"})
renov_ia = por_fuente_ia[por_fuente_ia["fuente_generacion"] == "Renovable"].merge(total_ia, on="fecha")
renov_ia["participacion"] = renov_ia["potencia_instalada_mw"] / renov_ia["total"] * 100

plt.plot(renov_ia["fecha"], renov_ia["participacion"])
plt.title("Participación de las renovables en la potencia instalada")
plt.xlabel("Período")
plt.ylabel("%")
plt.show()
````
