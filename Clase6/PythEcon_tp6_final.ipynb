## 1. Objetivo

Construir, con sus propios datos, un análisis **reproducible** que responda una **pregunta económica con una dimensión territorial**, y poder **explicar y defender cada decisión**.

Es el TP que integra el curso: van a obtener datos con código (clase 5), limpiarlos y validarlos con expresiones regulares (clase 4), integrarlos con pandas (clases 3 y 4), comunicarlos con gráficos (clases 3 y 4) y con mapas (clase 6), y dejarlos listos para que otra persona, o ustedes dentro de un año, los vuelva a generar con un solo comando.

**Qué se evalúa, en el fondo:** que el análisis sea correcto, que lo hayan **verificado** y que lo **entiendan**. Un TP más chico, verificado y bien explicado vale más que uno grande que no se puede ejecutar o que no pueden explicar.

### 1.1 Alcance

**Qué no alcanza:** una base que ya viene limpia (por ejemplo, de Kaggle) y solo se grafica; un mapa sin una pregunta detrás; código que no pueden explicar.

**Si sus datos no tienen geografía**, parte del trabajo es vincularlos a una capa geográfica (por código de provincia, departamento, municipio o país, o por coordenadas). Si de verdad no hay ninguna dimensión territorial posible, escríbannos antes de avanzar.

## 2. Requisitos mínimos

Cada requisito dice **qué** tiene que haber y **cómo nos damos cuenta** de que está. Los requisitos 3 a 6 tienen que estar resueltos **con sus datos**, no con datos de ejemplo.

### R1. Pregunta, unidad y período

Al principio del notebook principal y en el `README.md`: la pregunta (una o dos oraciones), la unidad de análisis (provincia, departamento, municipio, establecimiento, país…), el período y por qué la pregunta importa para su tesis o para la política económica.

### R2. Datos y obtención reproducible

- Al menos **dos fuentes distintas**, y **al menos una obtenida con código**: una API, *scraping* de HTML, extracción de PDF o descarga de archivos con `requests`.
- Los datos tal como se descargaron van en `datos/crudo/` y **nunca se editan a mano** (ni en Excel). Toda corrección se hace con código y queda registrada.
- Cada fuente queda documentada en el `README.md`: URL, quién la produce, fecha de descarga, licencia o condiciones de uso, y qué contiene.
- Las descargas usan `timeout`, verifican la respuesta (`raise_for_status()` o equivalente) y no vuelven a descargar lo que ya está (ver la sección 3).

### R3. Expresiones regulares

Al menos **dos usos sustantivos** de expresiones regulares sobre sus datos: extraer, validar o normalizar información (códigos, fechas, nombres, montos, textos de un PDF, URL…). Reemplazar una coma por un punto no cuenta.

Para cada uno:
- una función con *docstring* que explique qué hace y qué supuestos tiene;
- **casos de prueba** con `assert`: textos que deben coincidir (con el resultado esperado) y textos que **no** deben coincidir;
- cuántas filas coincidieron y cuántas no, y qué hicieron con las que no coincidieron.

Tomen como modelo la función `codigo_desde_texto` del desafío de la clase 6.

### R4. Integración con pandas

- **Tipos correctos**: códigos como texto (con sus ceros a la izquierda), fechas como fecha, números como número. Verifíquenlo con `dtypes`.
- Al menos **un `merge` verificado** entre fuentes distintas: con `validate=` e `indicator=True`, la cantidad de filas antes y después, y una explicación de las filas que no coincidieron (cuántas, cuáles, por qué y qué decidieron).
- Al menos una **agregación** con `groupby`.
- Idealmente un **cambio de forma** (`melt`, `pivot`, `pivot_table`) o la construcción de un **panel** (unidad × período).

### R5. Gráficos

Al menos **dos gráficos que no sean mapas** (matplotlib o seaborn), que respondan preguntas distintas. Cada uno con título, ejes con su unidad y fuente. El tipo de gráfico tiene que ser adecuado para lo que muestra, y debajo de cada uno, una o dos oraciones con lo que se ve.

### R6. Mapas

- Al menos **un mapa coroplético terminado**, con el cual respondan estas cinco preguntas: qué mide, con qué geometrías, en qué sistema de coordenadas, con qué escala y colores, y qué falta. Pueden usar y adaptar la función `mapa_coropletico` de la clase.
- Y **al menos uno** de estos:
  - (a) un mapa con puntos y una **unión espacial** (`sjoin` o `sjoin_nearest`);
  - (b) una **comparación** entre dos períodos, grupos o escenarios, con la misma escala en todos los paneles;
  - (c) un **mapa interactivo** (`explore()`) con `tooltip`, guardado como HTML.
- Si calculan áreas o distancias, en una proyección adecuada (igual área para áreas; en metros para distancias).
- La capa geográfica está documentada como una fuente más (R2), y la unión con sus datos es **por código** siempre que sea posible.

### R7. Verificación

- Al menos **cinco `assert` sobre sus datos** que verifiquen cosas que tienen que ser ciertas: claves únicas, totales que cierran, rangos posibles, cantidad de unidades esperada, ausencia de faltantes donde no puede haberlos, cantidad de filas después de cada `merge`…
- Una sección breve **«Qué verifiqué y qué no»**: qué controles hicieron (incluidos los que hicieron mirando), qué no pudieron verificar y qué podría estar mal.

### R8. Reproducibilidad y calidad del código

- Estructura de carpetas explicada en el `README.md`.
- `requirements.txt` con versiones fijas (`paquete==versión`), solo con lo que usan.
- `README.md` con la plantilla de la sección 6, que incluya **cómo ejecutar todo desde cero**.
- `.gitignore` con lo que no va al repositorio.
- Rutas relativas (nada de `C:\Users\...`), sin claves ni contraseñas en el código.
- Funciones para lo que se repite, nombres claros, celdas que se ejecutan en orden de arriba hacia abajo. Pueden usar la [guía de estilo PEP 8](https://recursospython.com/pep8es.pdf).

### R9. Conclusiones y limitaciones

Al final del notebook de análisis, entre 300 y 600 palabras: qué responden los datos a la pregunta, cuáles son las limitaciones (de los datos, de las uniones, de las decisiones que tomaron) y qué harían  con más tiempo, como por ej. para sus tesis. Un mapa muestra asociaciones en el espacio, no causas: no concluyan más de lo que los datos permiten.

### R10. Bitácora de uso de IA

Obligatoria, **aunque no hayan usado IA** (en ese caso, lo dicen). Ver la sección 5.

## 3. Datos pesados y reproducibilidad

Varios de ustedes van a trabajar con muchos archivos o con archivos grandes (por ejemplo, un JSON por municipio o registros administrativos de varios años). **No suban los datos crudos pesados al repositorio**: lo que se versiona es el código que los obtiene. Las reglas:

1. **Límite.** GitHub avisa con archivos de más de 50 MB y rechaza los de más de 100 MB. Si los datos crudos superan los 50 MB en total, la carpeta `datos/crudo/` va en el `.gitignore` y para corregir la generaremos con el script de obtención.
2. **Descarga incremental.** El script verifica si cada archivo ya existe antes de pedirlo, espera un poco entre pedidos (`time.sleep`), reintenta ante errores transitorios y guarda un registro de lo que falló. Si se corta a mitad de camino, retoma donde quedó.
3. **Modo muestra.** Incluyan en su código una variable `MODO_MUESTRA = True` hace que todo el flujo, desde la descarga al último mapa, corra con un subconjunto (por ejemplo, 10 municipios o un año) **en menos de 5 minutos**. Con `MODO_MUESTRA = False` corre todo, y el `README.md` dice cuánto tarda.
4. **Base final.** Si la base procesada pesa menos de 20 MB, súbanla en `datos/procesado/` (en Parquet o CSV comprimido) para que podamos revisar los resultados sin descargar todo.
5. **Claves de API.** Si usan una API que requiera claves nunca las guarden en el código ni en el repositorio. Guardenlas en un archivo `.env` (que va en el `.gitignore`), con instrucciones en el `README.md` para que los profes puedan conseguir una clave propia.
6. **Datos confidenciales** (por ejemplo, microdatos de la tesis con acuerdo de confidencialidad): avísennos antes del avance. Opciones: subir una muestra sintética con la misma estructura, trabajar con datos agregados o hacer el flujo completo con una fuente pública equivalente.

Un esquema de descarga incremental, para adaptar:

```python
import json
import time
from pathlib import Path

import requests

CRUDO = Path("datos/crudo/api_municipios")
CRUDO.mkdir(parents=True, exist_ok=True)


def descargar_municipio(codigo, reintentos=3):
    """Descarga el JSON de un municipio si todavía no está en disco. Devuelve la ruta o None."""
    destino = CRUDO / f"{codigo}.json"
    if destino.exists():
        return destino
    for intento in range(reintentos):
        try:
            respuesta = requests.get(URL_BASE, params={"municipio": codigo}, timeout=30)
            respuesta.raise_for_status()
            destino.write_text(json.dumps(respuesta.json(), ensure_ascii=False), encoding="utf-8")
            time.sleep(0.5)          # no saturar el servidor
            return destino
        except requests.RequestException as error:
            print(f"{codigo}: intento {intento + 1} falló ({error})")
            time.sleep(2 ** intento)
    return None


codigos = CODIGOS[:10] if MODO_MUESTRA else CODIGOS
```

## 5. Uso de IA

El objetivo del curso es que ustedes **sepan hacer** estas tareas y puedan **evaluar** el código, venga de donde venga. Por eso, en el TP Final la IA puede ser una tutora, no la autora.

| Permitido | No permitido |
| --- | --- |
| Pedir explicaciones de un concepto, de una función o de un mensaje de error | Pedir que genere el flujo completo, un notebook o la solución de un requisito (R3 a R7) para copiar |
| Consultar sintaxis o documentación | Delegar decisiones analíticas que no entienden (qué variable, cómo tratar faltantes, qué proyección, qué unión) |
| Pedir una revisión de **su** código y que ustedes decidan qué cambiar | Entregar código que no pueden explicar línea por línea |
| Mejorar la redacción del `README.md` o de las conclusiones | |

**La bitácora (R10).** Al final del `README.md`, una sección «Bitácora de uso de IA» con:

1. una tabla con una fila por uso relevante: herramienta, para qué la usaron, qué le pidieron (en una línea) y qué hicieron con la respuesta (la usaron tal cual, la modificaron, la descartaron);
2. al menos **un caso en que la IA se equivocó**, sugirió código desactualizado o algo que no correspondía a sus datos, y cómo se dieron cuenta (si la usaron);
3. si no usaron IA, una línea que lo diga.

**La defensa es la verificación.** Les podemos preguntar por cualquier línea del repositorio. Si no pueden explicar una parte, esa parte se evalúa como no entregada, sin importar quién la haya escrito.
