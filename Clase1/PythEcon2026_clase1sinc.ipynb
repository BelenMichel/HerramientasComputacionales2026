No pasa nada. De hecho, nos pasa seguido. Los Notebooks hace fácil aislar el error y revisar. Algunas cosas nos van a ayudar con los errores:

1.   **Leer el mensaje de error:** En particular, mirar qué tipo de error ocurrió, qué dice el mensaje y en qué línea aparece.
2.   **Revisar esa línea y el contexto inmediato:** Muchas veces el problema es algo simple: un nombre mal escrito, paréntesis/comillas que faltan, indentación o estar usando un tipo de dato distinto del esperado
3.   **Consultar la documentación:** Cuando no sabemos cómo debería comportarse una función u objeto: https://docs.python.org/3/tutorial/.
4.   **Buscar el problema:** Stack Overflow, GitHub Issues y otras comunidades siguen siendo muy útiles: probablemente alguien ya se encontró con un error parecido.
5.   **Usar IA como tutor de debugging:** Podemos mostrarle el error y nuestro código para pedirle que nos ayude a entender qué está pasando. Pero su propuesta también puede estar equivocada: hay que ejecutarla, revisarla y entenderla antes de aceptarla.

En Resumen:

Construyan un diccionario para cada integrante del equipo. Cada diccionario debe contener los siguientes keys:

Llamen `{nombre}_carac` a cada diccionario, donde `{nombre}` sea el nombre de cada integrante del equipo.

### (B) Bucle `for` (o *loop* `for`)

Construyan una lista con el apodo de cada integrante del grupo. Luego definan un for loop que itere por cada elemento de la lista y lo use como key en el diccionario **alumnos** para llamar al diccionario que contiene las caracteristicas de cada alumno. Finalmente, en cada iteración el loop debe imprimir:
* el nombre (no el apodo)
* el gusto de helado favorito

Tengan en cuenta que para imprimir van a tener que escribir código de la siguiente estructura:

`print("Texto que quieran escribir" + str(alumnos[apodo][key_diccionario_individual]))`

A tener en cuenta:
* `str()`: transforma el tipo de dato a un string
* `alumnos["X"]`: va a devolver el valor asociado a la llave "X", en este caso un diccionario individual
* `apodo`: va a ser uno de los elementos de la lista de apodos (en formato string)
* `[key_diccionario_individual]`: llama a un key del diccionario individual que nos devolvió `diccionario_apodos[X]`. La llave debería ser una de las variables que queremos imprimir

````python
apodos = ["dami_s", "dami_m", "lau"]
````

````python
for apodo in apodos:
  print("Nombre: " + alumnos[apodo]["nombre"])
  print("Helado Favorito: " + str(alumnos[apodo]["helado_favorito"]))
````

### (C) Loop `while`

Ahora construyan una variable `contador`. Luego usen esa variable para iterar sobre un while loop mientras que contador sea mayor a 0. En cada iteración se debe restar 1 del contador. En la clase asincrónica vimos una forma de hacer esto, otra forma es la siguiente:


````python
# corran esta celda
contador = 2

while contador >= 0:
    print(apodos[contador]) # imprime count antes de restar 1
    contador -= 1 # redefine la variable count tomando el valor actual - 1
````

Usando esa estructura, usen la variable count para indexar la lista con el apodo de cada integrante del grupo. Impriman:
* el apodo

Dentro del mismo loop, usen esa lista indexada para llamar al key del diccionario ya definido (igual que el ejercio anterior), pero ahora impriman:
* la edad

````python
apodos[0], apodos[1], apodos[2]
````

#### (5.1) Como armar listas de números con la función range()

Ya aprendimos a armar listas:

````python
num_list = [0, 1, 2, 3, 4, 5]
````

Para secuencias regulares de números (incrementos homogéneos), podemos usar la función range() para hacer lo mismo. Muy útil para loops.

````python
# La función tiene tres partes range(start, stop, step)
# para ver la documentación de una variable podemos hacer:
?range
````

````python
# Probemos
type(range(0,6))
````

Bueno, técnicamente no produce una lista, sino un _range object_ que es una secuencia de números sobre la cual un loop puede iterar. Pero, aplicando una función se puede castear a una lista:

````python
list(range(0,6))
````

Pero no es necesario hacerlo si lo quiero usar en un loop:

````python
for n in range(1,6):
    print(n)
````

Y ahora un ejemplo con pasos mayores a 1

````python
for n in range(10,100,20):
    print(n)
````
